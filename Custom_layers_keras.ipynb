## Creating Our Own neural network layers

```python
#To create a custom layers we need to inherit the Layer class from layers
from tensorflow.keras.layers import Layer
import tensorflow as tf

from Project_Classification_and_Captioning_Aircraft import summary


class MycustomLayer(Layer):
    #we need to implement three methods
    def __init__(self,units=32,**kwargs):
        super(MycustomLayer,self).__init__(**kwargs)
        self.units = units
    def build(self,input_shape):
        self.w = self.add_weight(shape=(input_shape[-1],self.units),initializer='random_normal',trainable=True)
        self.b = self.add_weight(shape=(self.units,),initializer='zeros',trainable=True)
    def call(self,inputs):
        return tf.matmul(inputs,self.w)+self.b
```
